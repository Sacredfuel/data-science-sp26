## Report


### Overview:

Why did you choose this dataset? What made you ask this question? These should all be answered in the overview section. Think of the overview section as the equivalent of an abstract in a research paper or an elevator pitch for the project. The following questions will help you frame your thoughts if you ever have to succinctly describe your project in an interview:

- **Title**: should capture the topic/theme of your project.
- **Objective**: In 1 to 2 sentences, succinctly describe what you are hoping to accomplish in this project in simple, non technical English.
- **Importance**: In 1 to 2 sentences, describe why this project has personal significance to you.
- **Originality**: In 1 to 2 sentences, describe why you believe this project idea is unique and original.

### Data:

The following questions will guide you through some criteria you should be using to assess if the data you have is enough for a successful project.

You'll need to indicate why your dataset links to the problem you asked in the overview, and prove why it is the appropiate choice for this problem.

- **Data Source**: Include a list of your data source(s), complete with URL(s) for downloading. All data must be publicly available.
- **Data Volume**: How many columns in your dataset? How many rows? If you are joining multiple datasets together, please tell us how many rows and columns remain after the data has been merged into a single dataset.
- **Data Richness**: What type of data is in your dataset? You don't need to describe every column. A generalized overview is fine. (e.g. "My data contains 311 complaint types, the date the complaints are created and closed, as well as a description of the complaint"). If you found a data dictionary, feel free to link us to that as well.

### The Predictive Model:

A strong data science project should demonstrate your knowledge of modeling. Indicate which model you chose and why it is the appropiate choice for this problem. You need to identify first whether your problem is Regression or Classification, and then you'll need to identify which model you'd like to use based on your problem type.


- **The Predictors (X's)**: Which column(s) in the dataset are being used to predict the column listed above?
- **Python Dependencies**: What Python libraries and dependencies are you be using?
- **Security and Privacy Considerations**: Are you be working with personal identifiable information (PII)? Can your model be mis-used for evil, not good? If so, how do you plan to mitigate that?

### The Visualization:

A key part of making a great data science portfolio are the visualizations. This is a quick and elegant way of showcasing your work during the job hunting process, even to a non-technical audience.

Thus, a major part of this final project will center around making the following types of visualizations with the data you choose. If your data cannot support these types of visualizations, then please, reconsider choosing another dataset.

- **Summary Statistics Plots**: Write out in detail at least 3 types of summary statistics graphs you plan to make with your data (e.g. "I plan to make a histogram using the column X").
- **Model Performance Plots**: At the time of writing this proposal, we would not have covered how to visualize model accuracy yet. So, no worries if this part is still confusing to you. Give it your best shot on explaining what kind of visualization you think will best showcase that your model is "successful" and "accurate".

### Results

Your results should speak for themselves. Provide the hard numbers and data that you found while working on the project and make sure to stick purely to the data. Currently you want to just present the findings you found while working on the data.

Generally your results should include summary statistics such as Loss and Model Accuracy but also some visualization of how close your model is to representing the real world scenario.

### Discussion

The discussion is your chance to explain just why your project is so meaningful. In this part, you can link your results to real world phonemena and try to answer your original question based on the results you found earlier. It's open ended and (generally) the most fun part of writing a paper since you can openly express how your project went along with the discoveries you made and what you would like to explore further.
